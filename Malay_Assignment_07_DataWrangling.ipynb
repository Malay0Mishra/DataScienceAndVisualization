```python
#Ques-28
original = pd.read_csv("marks_original.csv")
updated = pd.read_csv("marks_updated.csv")

original = original.set_index("Student_ID")
updated = updated.set_index("Student_ID")

result1 = original.combine_first(updated)


result2 = original.combine(updated, np.fmax)


comparison = pd.DataFrame({
    "combine_first": result1["Marks"],
    "np_fmax": result2["Marks"]
})

display(comparison)

different = comparison[
    comparison["combine_first"] != comparison["np_fmax"]
]

print("Student IDs where results differ:")
display(different)
```

```python
#Ques-29

long = pd.read_csv("marks_long.csv")


wide = long.pivot(
    index="Student_ID",
    columns="Subject",
    values="Marks"
)


stacked = wide.stack()

print("Stacked DataFrame:")
display(stacked)

restored = stacked.unstack()

print("Restored Pivot Table:")
display(restored)
```

```python
#Ques-30

merged = pd.merge(
    students,
    long,
    on="Student_ID",
    how="inner"
)


wide = merged.pivot(
    index=["Student_ID", "Student_Name"],
    columns="Subject",
    values="Marks"
)

print("Pivot Table:")
display(wide)


result = wide.reset_index()

print("Final DataFrame:")
display(result)
```
