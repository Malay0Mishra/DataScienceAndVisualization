```python
#Question -3(TOPIC 8)
counter = 0

def increment_counter():
    counter = counter + 1

increment_counter()
#Python treats counter as a local variable because you're assigning to it inside the function, but you're trying to read it before assigning it.
```

```python
#Question -4(TOPIC 8)
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
